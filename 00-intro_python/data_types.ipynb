## Variable

Para iniciar o desenvolvimento em python, podemos compreender a estrutura de dados em variáveis (tipos de dados)

```python
var_int = 1
type(var_int) #aqui obtemos informações do tipo de variável inteiro

var_float = 0.1
type(var_float) # variável float

var_str = 'Hello Word!'
type(var_str) # variável string
```

É interessante notar que as variáveis não podem assumir qualquer tipo de nome, por exemplo palavras que são usadas em para alguma operação ou sentença lógica (if, else, not, True, False...)

```python
# Se quisermos de modo geral obter informações do projeto com todas as variáveis e seus tipos, usamos o comando:

# obtemos de forma geral as variáveis, tipos e data/info
%whos 

# apresenta as variáveis em lista, podemos usar %who também.
%who_ls 
```

```python
# multiple assignment

#podemos criar variáveis numa mesma linha, porém não recomendado

varA, varB = 3, 'test'

print(varA)
print(varB)

#Podemos reescrever as variáveis existentes (overwrite) atribuindo novos valores a elas. Note que é possível ainda mudar seu tipo, sem problemas.

varA = 20
varB = 0.1

print(varA)
print(varB)
```

## Math

Vamos abordar de maneira geral as operações matemáticas padrões do python

```python
x = 7
y = 4.3
z = 2

y+x - 7 #podemos realizar operações com float e int, o resultado final é float

#realizando múltiplas operações
# * mult
# / divid
# ** power
# - menos and + mais

result = (y + 3.5 * x - 7)/z

result
```

```python
# É interessante que o python pode concatenar strings, por exemplo:

test1 = 'Igor'
test2 = ' Oliveira'

#overloading
name = test1 + test2
name

# posso ainda fazer multiplicações para inteiros.

name = test1*3
name

# para prints, temos que tomar o cuidados de transformar as variáveis em str:

print(str(x) + ' + ' + str(y) + ' = ' + str(x+y)) #temos inteiro e string, por isso preciso converter para str.

print(x+y) # só que aqui temos apenas um tipo de dado (int)
```

```python
# podemos dar valores às variáveis enquanto o programa roda usando o input:

numb = float(input('Give me a number: '))

type(numb) # só que note, puxando o input direto, temos como resultado uma str, caso seja de interesse fazer alguma operação matemática, precisamos converter.

result = numb * 10
result
```

```python
# Para o teorema de Pitágoras, usando três lados com valores inteiros a,b e c.

a = int(input("Valor a: "))
b = int(input("Valor b: "))

result = (a**2+b**2)**(1/2)
result
```

## Listas

Vamos ver algumas ideias sobre listas

```python
# Podemos chamar uma list da seguinte forma:

aList = [23, 0.2, 'legal', [34,'-i']] # as listas podem conter diferentes diferentes tipos de atributos, inclusive listas de listas.
type(aList)

# podemos fazer verificações nas listas, por exemplo:

23 not in aList # no caso é False, pois o 23 está na lista!
0.2 in aList # no caso é True, pois temos o 0.2 na lista.

# podemos escrever as listas como:
aList = [23,
        0.2,
        'legal',
        [34,'-i']]

# Podemos tambem fazer somas de elementos da listas:

list1 = [1,2,3]
list2 = ['a','b','c']
list1+list2

# ou ainda multiplicações

list1 * 3
```

```python
# podemos adicionar ou retirar elementos da lista

list1 = [1,2,3, -100, 0.2]

list1.append(-100)
print(list1)
# podemos inserir elementos numa posição específica:

list1.insert(2,9)
print(list1) # inserimos o número 9 na posição 2.

# podemos remover um valor específico

list1.remove(9)
print(list1)

# podemos fazer uma organização da lista do menor para o maior

list1.sort() # note que o sort só funciona para float e int, dá erro em str
list1

# Caso seja de interesse, podemos selecionar elementos específicos pela sua posição:

list1[2]

# Se quisermos podemos verificar quantos elementos com determinado valor na lista

list1.count(-100)
print(list1)

# podemos modificar a lista, colocando seus elementos de trás pra frente

list1.sort(reverse=True)
print(list1)

# indexação, substituindo um elemento:

list1[0] = 13.13
print(list1)
```

## Boolean

O tipo boolean aceita apenas dois elementos True or False

```python
# boolean

t = True
f = False

%whos

# usamos os booleanos para compreender se uma sentença está correta ou não, por exemplo:

4==4 # é True
4==3 # é falso...

# assim podemos realizar operações lógicas:

4>4 # falso, pois 4 é igual a 4
456 >= 21 # verdeiro!

3<20 # verdadeiro
3<=2 # verdadeiro

22 != 14 # verdadeiro, é diferente

# podemos ainda realizar várias comparações lógicas usando o and e or:

22 != 14 and 2 >3 # falso, apesar de 22 ser diferente de 14, 2 é menor que 3

22 != 14 or 2 >3 # verdadeiro, pois aqui só precisa que apenas uma sentença seja verdadeira (22 diferente de 14)
```

```python
# no python o valor bool(0) é único falso, enquanto que bool(!=0) é verdadeiro

bool(0) #false
bool('1312') #true

# Se quisermos 3 número para verificar seus lados se são triângulo retângulo:

a = float(input('Digite a'))
b = float(input('Digite b'))
c = float(input('Digite c'))

c == (a**2+b**2)**(1/2)
```

## Dictionaries

Dicionários são parecidas com listas, porém a gente atribui os índices específicos aos elementos. É um dados supervisionados de modo geral.

```python
# dictionaries: key/value pairs

dic = {'nome':'igor', 'sobrenome':'oliveira'}
type(dic)

%whos

# podemos atribuir um elemento, como não especificado sua posição, ele entra no final:

#     key      value
dic['name'] = 'Test'
dic

dic['ageRange'] = [20,50]
dic

# podemos por exemplo recuperar um elemento a partir da sua chave:

dic['nome']

# podemos ver quais chaves existem no dicionário:

dic.keys()

# e fazer verificações lógicas

'name' in dic # note que isso faz sentido apenas para keys, não para elementos

# para obter informações dos valores, fazemos:

dic.values()

# ou ainda se quisermos informações do itens (key/elements)

dic.items() # isso mostra o par de chave e elemento
```

```python
# vamos por um input para dois números

numb1 = float(input('digite um número: '))
dic['firtNum'] = numb1
numb2 = float(input('digite um número: '))
dic['secondNum'] = numb2

dic['product'] = numb1*numb2
# e criar um dicionário usando três chaves

# outra forma é:
#dic = {'firtNum':numb1, 'secondNum':numb2, 'product':numb1*numb2}
dic
```
