## Variable

Para iniciar o desenvolvimento em python, podemos compreender a estrutura de dados em variáveis (tipos de dados)

```python
var_int = 1
type(var_int) #aqui obtemos informações do tipo de variável inteiro

var_float = 0.1
type(var_float) # variável float

var_str = 'Hello Word!'
type(var_str) # variável string
```

É interessante notar que as variáveis não podem assumir qualquer tipo de nome, por exemplo palavras que são usadas em para alguma operação ou sentença lógica (if, else, not, True, False...)

```python
# Se quisermos de modo geral obter informações do projeto com todas as variáveis e seus tipos, usamos o comando:

# obtemos de forma geral as variáveis, tipos e data/info
%whos 

# apresenta as variáveis em lista, podemos usar %who também.
%who_ls 
```

```python
# multiple assignment

#podemos criar variáveis numa mesma linha, porém não recomendado

varA, varB = 3, 'test'

print(varA)
print(varB)

#Podemos reescrever as variáveis existentes (overwrite) atribuindo novos valores a elas. Note que é possível ainda mudar seu tipo, sem problemas.

varA = 20
varB = 0.1

print(varA)
print(varB)
```

## Math

Vamos abordar de maneira geral as operações matemáticas padrões do python

```python
x = 7
y = 4.3
z = 2

y+x - 7 #podemos realizar operações com float e int, o resultado final é float

#realizando múltiplas operações
# * mult
# / divid
# ** power
# - menos and + mais

result = (y + 3.5 * x - 7)/z

result
```

```python
# É interessante que o python pode concatenar strings, por exemplo:

test1 = 'Igor'
test2 = ' Oliveira'

#overloading
name = test1 + test2
name

# posso ainda fazer multiplicações para inteiros.

name = test1*3
name

# para prints, temos que tomar o cuidados de transformar as variáveis em str:

print(str(x) + ' + ' + str(y) + ' = ' + str(x+y)) #temos inteiro e string, por isso preciso converter para str.

print(x+y) # só que aqui temos apenas um tipo de dado (int)
```

```python
# podemos dar valores às variáveis enquanto o programa roda usando o input:

numb = float(input('Give me a number: '))

type(numb) # só que note, puxando o input direto, temos como resultado uma str, caso seja de interesse fazer alguma operação matemática, precisamos converter.

result = numb * 10
result
```

```python
# Para o teorema de Pitágoras, usando três lados com valores inteiros a,b e c.

a = int(input("Valor a: "))
b = int(input("Valor b: "))

result = (a**2+b**2)**(1/2)
result
```
