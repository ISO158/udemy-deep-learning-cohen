## Printing and string interpolation

```python
num = 11.7
sng = 'cellar doors'

print('I would like to have ' + str(num) + ' ' + sng)

# podemos usar outras maneiras compactas para expressar essas ideias.
```

```python
print('I would like to have %g %s.' %(num,sng))
# g = float, s = str, i = int

print('I would like to have %.3f %s.' %(num,sng)) 
# podemos usar o .f para especificar a quantidade de números decimais
```

```python
# f string

print(f'I would like to have {num:.2f} {sng}.') 
```

```python
print('I would like to have {n:.2f} {s}.' .format(n=num,s=sng)) 
```

## Subplot geometry

```python
import matplotlib.pyplot as plt
import numpy as np

plt.subplot(2,2,1)
plt.plot(np.random.randn(10))

plt.subplot(2,2,4)
plt.plot(np.random.randn(10))
```

```python
fig,ax = plt.subplots(1,2,figsize=(10,5))
x = np.arange(10)
ax[0].plot(x,x**2,'b')
ax[0].plot(x,x,'k')

ax[1].plot(x,np.sqrt(x),'r')

plt.show()
```

```python
fig,ax = plt.subplots(2,2)

ax[0,0].plot(np.random.randn(4,4))
ax[0,1].plot(np.random.randn(4,4))
ax[1,0].plot(np.random.randn(4,4))
ax[1,1].plot(np.random.randn(4,4))

plt.tight_layout()
plt.show()
```

```python
# Exercise: create a 3x3 subplot geometry 
# in a for loop

fig,ax = plt.subplots(3,3,figsize=(7,7))

for i in range(3):
    for j in range(3):
        ax[i,j].plot(np.random.randn(4,4))

plt.tight_layout()
plt.show()

M = np.random.randint(10,size=(4,4))
print(M),print(' ')
print(M.flatten())
```

```python
fig,ax = plt.subplots(3,3,figsize=(7,7))

for i in ax.flatten():
    i.plot(np.random.randn(3,9))

plt.show()
```

## Making the graphs look nicer

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-3,3,101)

plt.plot(x,x,label='y=x')
plt.plot(x,x**2,label='y=x**2')
plt.plot(x,x**3,label='y=x**3')

plt.legend()
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('A really awesome plot.')
plt.xlim([x[0],x[-1]])
plt.ylim([-10,10])

plt.gca().set_aspect('auto') #podemos usar .3

plt.show()
```

```python
fig,ax = plt.subplots(2,3,figsize=(15,7))

ax[0,0].plot(x,x,label='y=x')
ax[0,1].plot(x,x**2,label='y=x**2')
ax[0,2].plot(x,x**3,label='y=x**3')

# ax.legend()
# ax.set_xlabel('x')
# ax.set_ylabel('f(x)')
# ax.set_title('A really awesome plot.')
# ax.set_xlim([x[0],x[-1]])
# ax.set_ylim([-10,10])

plt.gca().set_aspect('auto') #podemos usar .3
plt.tight_layout()
plt.show()
```

## Seaborn

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import seaborn as sns
```

```python
n = 200
D = np.zeros((n,2))
D[:,0] = np.linspace(0,10,n) + np.random.randn(n)
D[:,1] = D[:,0]**2 + np.random.randn(n)*10

sns.jointplot(x=D[:,0],y=D[:,1])

plt.show()
```

```python
# podemos converter esse conjunto de dados em um dataframe

df = pd.DataFrame(data=D,columns=['var1','var2'])

# estamos modificando o nome dos eixos
sns.jointplot(x=df.columns[0],y=df.columns[1],data=df, kind='hist',color='purple')
plt.show()
```
