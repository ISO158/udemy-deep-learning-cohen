## Printing and string interpolation

```python
num = 11.7
sng = 'cellar doors'

print('I would like to have ' + str(num) + ' ' + sng)

# podemos usar outras maneiras compactas para expressar essas ideias.
```

```python
print('I would like to have %g %s.' %(num,sng))
# g = float, s = str, i = int

print('I would like to have %.3f %s.' %(num,sng)) 
# podemos usar o .f para especificar a quantidade de números decimais
```

```python
# f string

print(f'I would like to have {num:.2f} {sng}.') 
```

```python
print('I would like to have {n:.2f} {s}.' .format(n=num,s=sng)) 
```

## Seaborn

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import seaborn as sns
```

```python
n = 200
D = np.zeros((n,2))
D[:,0] = np.linspace(0,10,n) + np.random.randn(n)
D[:,1] = D[:,0]**2 + np.random.randn(n)*10

sns.jointplot(x=D[:,0],y=D[:,1])

plt.show()
```

```python
# podemos converter esse conjunto de dados em um dataframe

df = pd.DataFrame(data=D,columns=['var1','var2'])
# estamos modificando o nome dos eixos

sns.jointplot(x=df.columns[0],y=df.columns[1],data=df, kind='hist',color='purple')
plt.show()
```

```python
x = np.linspace(-1,1,n)
y1 = x**2
y2 = np.sin(3*x)
y3 = np.exp(-10*x**2)

sns.scatterplot(x=y1,y=y2,hue=y3,palette='Spectral')
plt.show()
```

```python
# exercise: create a regression plot of data in matrix D -> df

n= 500

mat = np.zeros((n,2))
mat[:,0] = np.linspace(0,1,n) + np.random.randn(n)
mat[:,1] = mat[:,0]*2 + np.random.randn(n)*10

df = pd.DataFrame(data=mat,columns=['Var1','Var2'])

# estrutura para realizar a regressão dos dados
sns.regplot(x=df.columns[0],y=df.columns[1],data=df,color='green')
plt.title(f'Regression of {df.columns[0]} on {df.columns[1]}')
plt.show()
```

$$
H_{ij} = \frac{1}{i+j-1}
$$

```python
# Exercise: visualize the hilber matrix using seaborn (heatmap)
# H is 10x10 matrix
n = 10
H = np.zeros((n,n))

for i in range(n):
    for j in range(n):
        H[i,j] = 1/((i+1)+(j+1)-1)

sns.heatmap(data=H, vmax=0.4)
```
