## Operado try and except

```python
# dentro do python podemos usar a expressão try e except 

x = 'igor'

try :
    x = int(x) # note, como a variável x é uma string, o python não consegue converter ela para inteiro, então ele vai gerar um erro.
except :
    x = -1 # o bloco except vai ser executado, recebendo o valor -1.

print(x)
```

```python
# Essa estrutura de try e except é muito útil para tratar erros que podem ocorrer durante a execução do código, permitindo que o programa continue rodando mesmo após um erro. 

# Um exemplo clássico é o usuário digitar um valor inválido, como uma letra, quando o programa espera um número.

import math as math

try :
    y = int(input('Enter a number: '))
    result = math.factorial(y)
    print(f"The factorial of {y} is: {result}") # Esse print mostra o uso do f-string, formatando a string com valores de variáveis entre {}.
except ValueError:
    print("You did not enter a valid number.")
```

## If-else-statement

```python
# import the libraries

import numpy as np
```

```python
# if statements

x = 4

# conditional 

if x==4:
    print('ok!')
```

```python
r = np.random.randn()

if r>0:
    outcome = 'positive'
else:
    outcome = 'negative'

print('The number is ' + outcome)
```

```python
caketype = 'carrot'

if caketype == 'carrot':
    print('Total yum!')
elif caketype == 'chocolate':
    print('Yummy as well')
elif caketype == 'fruitcake':
    print('is this thing good for me?')
else:
    print("ok, i'll try it")
```
