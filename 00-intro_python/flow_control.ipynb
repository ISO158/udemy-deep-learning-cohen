## Operado try and except

```python
# dentro do python podemos usar a expressão try e except 

x = 'igor'

try :
    x = int(x) # note, como a variável x é uma string, o python não consegue converter ela para inteiro, então ele vai gerar um erro.
except :
    x = -1 # o bloco except vai ser executado, recebendo o valor -1.

print(x)
```

```python
# Essa estrutura de try e except é muito útil para tratar erros que podem ocorrer durante a execução do código, permitindo que o programa continue rodando mesmo após um erro. 

# Um exemplo clássico é o usuário digitar um valor inválido, como uma letra, quando o programa espera um número.

import math as math

try :
    y = int(input('Enter a number: '))
    result = math.factorial(y)
    print(f"The factorial of {y} is: {result}") # Esse print mostra o uso do f-string, formatando a string com valores de variáveis entre {}.
except ValueError:
    print("You did not enter a valid number.")
```

```python
# Exercise
# Create a function that computes the 'dot product' between two vectors
# 1) check the both inputs are numpy arrays (isinstance())
# 2) check that the two inputs are the same length
# 3a) break out of the function and give a helpful error message.
# raise Exception().
# 3b) Compute and return the dot product

import numpy as np

vec_a = np.array([1,2])
vec_b = np.array([4,9])
result = [0,0]

def dot_p(a,b):
    produto = int(np.dot(a,b)) # or we can using sum(a*b)
    
    if type(a)==type(np.array(a)) and type(b)==type(np.array(b)):
        print('ok, both inputs are numpy arrays')
    else:
        print('Inputs are not numpy arrays')

    if len(a) == len(b):
        print('Same length')
    else:
        print('It is different')

    return produto

dot_p(vec_a,vec_b)
```

```python
# podemos também verificar se o objeto é do tipo np.array usando o isinstance()

# isinstance(vec_a,np.ndarray) 
# type(vec_a) == type(np.array(vec_a))

def dotproduct(a,b):

    if not ( isinstance(a,np.ndarray) and isinstance(b,np.ndarray) ):
        raise Exception('Input must be numpy arrays')

    if not len(a)==len(b):
        raise Exception('Inputs must be the same lengths.')
    
    # return the dot product
    return sum(a*b)
```

```python
vec_a = np.array([1,2])
vec_b = np.array([4,9])

dotproduct(vec_a,vec_b)
```

## For loops

```python
# for loops

import numpy as np

numbers = np.linspace(3,17.1,22) # lembre-se o linspace é um conjunto de número com n espaçamento, iniciando em x e terminando em y (x,y,n).

for i in range(len(numbers)): # usando o range
    x = numbers[i]
    print(x)

for i in numbers: # podemos obter os elementos de numbers fazendo o for
    print(i)
```

```python
# Exercise
# Fibonacci series - starting [0,1,...,i+(i-1)]
# Create a function that reports the Fib. series fo input N.

import numpy as np

def fib_series(n):
    list_array = []
    for i in range(n):
        if i <= 1:
            list_array.append(i)
        else:
            list_array.append(list_array[i-1]+list_array[i-2])
    
    return list_array

fib_series(7)
```

```python
def fibseq(N):
    v = [0,1]
    for i in range(N-2):
        v.append( sum(v[-2:]) )

    return v

fibseq(7)
```

## Enumerate and zip

```python
# enumerate and zip

import numpy as np

ns = np.linspace(-5,5,7)

for i,n in enumerate(ns):
    # nessa estrutura temos que n = ns[i]
    print('Indes ' + str(i) + ' has a value of ' + str(n))
```

```python
text = 'Hello my name is Igor'

vowels = np.zeros(len(text))

for i,l in enumerate(text):
    if l in 'aeiou':
        vowels[i] = 1

vowels
```

```python
listA = [3,4,17]
listB = ['q','w','e']

for i in range(len(listA)):
    print(str(listA[i]) + ' ' + listB[i])
```

```python
# podemos realizar o mesmo processo usando o comando zip

listA = [3,4,17]
listB = ['q','w','e']

for a,b in zip(listA,listB):
    # é como se estivéssemos usando
    # a = listA[i]
    # b = listB[i] em um for i in range(len(listA)) por exemplo
    print(str(a) + ' ' + b)
```

```python
# O interessante do zip é que ele corta o for quando o número de elementos em ambos as listas é diferente.

listA = [3,4,17,32,45,1,0]
listB = ['q','w','e']

for a,b in zip(listA,listB):
    print(str(a),b) # só vai até o 3 elemento, devido a listB
```

```python
# Exercise: given the following lists, create a dictionary using zip

names = ['alpha','beta','gamma']
values = [10,20,30]
dic = {}

for n,v in zip(names,values):
    dic[n] = v
    
print(dic)

# uma maneira de usar o zip rapidamente é

D = dict(zip(names,values))
print(D)
```
