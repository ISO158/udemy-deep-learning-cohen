## Indexing

O indexing é utilizado para atribuir os dados aos índices, podemos aplicar isso nas listas, listas de listas, dicionários. A seguir alguns exemplos de aplicações:

```python
# index

alist = [4,1,8,-6, 'test', 0.98]

alist #eu posso mostrar toda lista

# se eu quiser selecionar o quarto elemento

alist[3]

# caso eu queira roda o último elemento
# eu posso usar o len(alist)-1 ou ainda -1

alist[len(alist)-1]
alist[-1]

# Se tivermos uma lista de lista ou lista com dicionários, por exemplo

alist = [4,1,[10,20,30],{'Test1':'olá','Test2':'Hello'}]

# selecionamos seus elementos

alist[2][1] # para selecionar a lista na posição 2 e elemento 1 (valor 20)

# no caso do dicionário é a mesma coisa, porém usamos seu índice atrelado.

alist[3]['Test1']
```

## Slicing

Usamos o slicing selecionar diferente elementos de uma lista, adicionando limite inferior e superior.

```python
#slicing

a = list(range(5,11))
a
```

```python
# se quisermos selecionar os primeiros dois elementos, usamos
a[0:2]
# note que a saída não inclui o elemento 7, da mesma forma que o range não inclui o valor 11.

# portanto se quisermos incluir o 7, precisamos. Note que podemos omitir o número 0, caso seja necessário apenas um limite superior.
a[:3]

# para ambos os limites, por exemplo queremos selecionar 6 e 7, fica:
a[1:3]

#se não queremos um limite superior, é a mesma ideia. Podemos selecionar todos os elemento a partir do 7, assim começamos a partir do índice 3
a[3:]
```

```python
# se quisermos mostrar os valores espaçados a acada 2 elementos, por exemplo, podemos fazer:
a = list(range(5,11))
a[::2]

#se quisermos a cada 3 elementos

a[::3]

# podemos adicionar também, onde começamos a contar a posição dos elementos

a[1::2]

# ou por exemplo, começamos no primeiro elemento e pulamos 4 depois

a[1::4]

#De maneira geral o python compreende a estrutura do slicing da seguinte forma:

#a[start:stop:skip]

#assim, o primeiro valor refere-se ao limite inferior, o próximo ao limite superior. Enquanto que o último diz quantas posições/elementos deve ser pulado.
```

```python
start = 0
stop = 3
skip = 2
a[start:stop:skip]
```

```python
# reverse a list
# e se quisermos reverter uma lista? Ou seja vamos começar em 10,9,...,5.

a[::-1]
#basicamente pegamos a lista toda :: e pulamos -1 índice.
```

```python
#Exercise

text = '345dfyug ecin a si ekiM 93845d'
# usar o slicing para escrever "Mike is a nice guy" do 'text'

# O slicing não é aplicado apenas em listas, mas pode ser usado em diferentes contextos, como uma string.

start = -8
stop = 4
skip = -1
text[start:stop:skip]
```
