## Functions

Compreender inputs and outputs, a sintaxe de funções em python

```python
# Uma função do python é a sum, que faz a soma de elementos, por exemplo de uma lista:

sum([10,20,30])

list = [20,20,20,7.0]
sum(list) # sum faz apenas soma de elementos int e float

# Note que a função possui o input, neste caso a lista, e um output, resultado da soma.

input  = list
output = sum(input)
print('The sum is '+ str(output))
```

```python
# Podemos criar as próprias funções, a partir de operações lógicas ou outros artifícios.

# Por exemplo, calcular a média da list

average = sum(list)/len(list)
print('The mean (average) of '+str(list)+' is '+str(average))
```

## Python libraries (numpy)

Existem bibliotecas que fornecem funções já setadas, assim chamando a library é possível usar seus recursos.

```python
# Podemos adicionar as bibliotecas ao projeto em ambiente virtual
# Abrimos o terminal
# python -m venv venv
# .venv\Scripts\Activate
# pip install "library"

# Para importar a biblioteca:

import numpy as np # em que a biblioteca é numpy e para chamarmos no nosso projeto usando uma abreviação: np

input = [10,20,30]

output = np.average(input)
print("Exemple of output using numpy function (average): " + str(output))
```

```python
# Uma função que podemos usar:

x = np.linspace(1,7,5) # temos 5 números linearmente espaçados entre 1 e 7.
type(x) # O resultado é uma matriz numpy
```

```python
input_np = np.array(input)

print(type(input)) # os dados originais são uma lista
print(type(input_np)) # quando usamos a função array convertemos para uma matriz numpy.
```
