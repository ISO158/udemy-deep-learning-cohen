## Functions

Compreender inputs and outputs, a sintaxe de funções em python

```python
# Uma função do python é a sum, que faz a soma de elementos, por exemplo de uma lista:

sum([10,20,30])

list = [20,20,20,7.0]
sum(list) # sum faz apenas soma de elementos int e float

# Note que a função possui o input, neste caso a lista, e um output, resultado da soma.

input  = list
output = sum(input)
print('The sum is '+ str(output))
```

```python
# Podemos criar as próprias funções, a partir de operações lógicas ou outros artifícios.

# Por exemplo, calcular a média da list

average = sum(list)/len(list)
print('The mean (average) of '+str(list)+' is '+str(average))
```

## Python libraries (numpy)

Existem bibliotecas que fornecem funções já setadas, assim chamando a library é possível usar seus recursos.

```python
# Podemos adicionar as bibliotecas ao projeto em ambiente virtual
# Abrimos o terminal
# python -m venv venv
# .venv\Scripts\Activate
# pip install "library"

# Para importar a biblioteca:

import numpy as np # em que a biblioteca é numpy e para chamarmos no nosso projeto usando uma abreviação: np

from numpy import mean as mn # dentro da biblioteca numpy podemos importar apenas uma função específica. Isso é interessante para não pesar o projeto.

# nesse caso, além de chamar a função, eu dou um apelido para ela

input = [10,20,30]

mn(input) # obtenho aqui a média dos elementos no input

output = np.average(input)
print("Exemple of output using numpy function (average): " + str(output))
```

```python
# Uma função que podemos usar:

x = np.linspace(1,7,5) # temos 5 números linearmente espaçados entre 1 e 7.
type(x) # O resultado é uma matriz numpy
```

```python
input_np = np.array(input)

print(type(input)) # os dados originais são uma lista
print(type(input_np)) # quando usamos a função array convertemos para uma matriz numpy.
```

```python
# Há funções que podemos aplicar direto aos objetos

input_np.mean() # média
input_np.min()  # mínimo
input_np.max()  # input

# podemos aplicar as funções também:

x = np.max(input_np) # 'origem da função' . 'função' (objeto)
x = input_np.max() # objeto . 'função direta do np'

# Ambos os métodos usados geram o mesmo resultado e mesmo tipo de var.
```

```python
# in-placa
# out-of-place / no-in-place operations

# Algumas operações no python alteram a variável chamando uma certo função. Por exemplo:

n = [4,3,5,2,6,1,7]
nn = np.array(n)

print(nn)
nn.sort() # a aplicação do método sort() altera o nn
print(nn) # variável nn está organizada


# (Esse é um exemplo de in-placa)
```

```python
# Porém se realizarmos o mesmo processo, porém usando a função do numpy
n = [4,3,5,2,6,1,7]
nn = np.array(n)

print(nn)
np.sort(nn) # a aplicação do método do np.sort() não altera o nn
print(nn)   # variável nn não sofre alteração.

# Para isso ser efetivo, precisamos atribuir a mudança realizada pelo método à variável:

nn = np.sort(nn) # por exemplo
print(nn)

# (Esse é um exemplo de no-in-placa ou out-of-place)
```

## Python libraries (Pandas)

Pandas é uma biblioteca que permite manipular tabelas (matrizes, listas, dicionários)

```python
import numpy as np
import pandas as pd
# os apelidos dados, np e pd, são módulos adicionados ao projeto.

# create random data
#np.random.randn(5,5) # criamos uma matriz 5x5 

var1 = np.random.randn(100)*5 + 20 # criamos uma matriz que apresenta 100 elementos distribuídos ao longo de 20 linhas e 5 colunas.

var2 = np.random.randn(100)>0 # podemos criar uma matriz que avalia cada elemento logicamente.

# variables labels
labels = ['Temp(C)', 'Ice cream']

# put them together into a dictionary
D = {labels[0]:var1, labels[1]:var2}

# import the dictionary into a pandas dataframe (df)

df = pd.DataFrame(data=D)
# O dataframe permite visualizar o conjunto de dados
D
```

```python
# Podemos visualizar partes dos dados:

df.head() # mostra os primeiros 5 dados

df.count() # mostra a quantidade de elementos para cada coluna, além do tipo de dados.

df.mean() # a média dos elementos no dicionário para cada coluna. Note que ele faz a média até de dados booleans, convertendo falsos em 0 e verdadeiros em 1.
```
