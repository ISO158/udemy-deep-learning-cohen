## Functions

Compreender inputs and outputs, a sintaxe de funções em python

```python
# Uma função do python é a sum, que faz a soma de elementos, por exemplo de uma lista:

sum([10,20,30])

list = [20,20,20,7.0]
sum(list) # sum faz apenas soma de elementos int e float

# Note que a função possui o input, neste caso a lista, e um output, resultado da soma.

input  = list
output = sum(input)
print('The sum is '+ str(output))
```

```python
# Podemos criar as próprias funções, a partir de operações lógicas ou outros artifícios.

# Por exemplo, calcular a média da list

average = sum(list)/len(list)
print('The mean (average) of '+str(list)+' is '+str(average))
```
