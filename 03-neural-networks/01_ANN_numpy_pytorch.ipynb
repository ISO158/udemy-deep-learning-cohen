### ANN Math

Basicamente a ANN é formada por um perceptron que representa o modelo que relaciona uma entrada e devolve saída:
$$
 x \rightarrow \hat{y}
$$

Por exemplo, podemos ter entradas $x_i$ com pesos $w_i$, sendo que $x_0$ é o termo de viés (biers), que são recebidos por uma função $\sum$ linear. Sua saída é recebida pela função não linear, chamada de função de ativação que por fim fornece o valor de $\hat{y}$.

É interessante notar que podemos escrever a relação entre a entrada e saída de forma matemática por:

$$
\hat{y} = \sigma (x_0 w_0 + \sum x_i w_i)
$$

onde o temo do somatório é dado pelo produto escalar entre $x^T w$, além disso o biers é definido como 1, por tanto:

$$
\hat{y} = \sigma(x^T w)
$$

notando que o termo $w_0$ foi absorvido pelo produto escalar.
