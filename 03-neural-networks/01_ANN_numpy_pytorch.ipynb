### ANN Math

Basicamente a ANN é formada por um perceptron que representa o modelo que relaciona uma entrada e devolve saída:
$$
 x \rightarrow \hat{y}
$$

Por exemplo, podemos ter entradas $x_i$ com pesos $w_i$, sendo que $x_0$ é o termo de viés (biers), que são recebidos por uma função $\sum$ linear. Sua saída é recebida pela função não linear, chamada de função de ativação que por fim fornece o valor de $\hat{y}$.



É interessante notar que podemos escrever a relação entre a entrada e saída de forma matemática por:

$$
\hat{y} = \sigma (x_0 w_0 + \sum x_i w_i)
$$

onde o temo do somatório é dado pelo produto escalar entre $x^T w$, além disso o biers é definido como 1, por tanto:

$$
\hat{y} = \sigma(x^T w)
$$

notando que o termo $w_0$ foi absorvido pelo produto escalar.

### Activation function

A função de ativação usa como entrada a saída da função anterior (ex.: somatório), aplicando uma análise não linear que pode ser:
 - Sigmoid
 - Hyperbolic tangent
 - ReLU
Apesar de existirem outas, elas são baseadas nessas três principais.

Por exemplo, a função Sigmoid transforma o range de dados do output da soma em valores que variam entre 0 e 1 (uma normalização). Essa função faz com que os valores tendam a 0 ou 1. 

A função tangente hiperbólica realiza de modo similar, porém entre -1 e 1, note que como a variação é maior em relação a função Sigmoid, temos uma descrição mais acentuada dos valores.

Já a função ReLU torna nulo todos os valores de entrada negativo, enquanto que valores positivos começam a crescer rapidamente de maneira linear.



Isso facilita a compreensão, por exemplo, do modelo em diferenciar categorias, já que há uma delimitação bem estrita.

Um dos pontos essenciais das redes neurais é o aprendizado por geração, ajustando os pesos e minimizando o erro do modelo, a partir de uma amostra de validação.

### ANN math - errors, loss and cost

O nosso modelo será construído para representar a realidade (conjunto de dados), mas não é garantido que isso seja 100% verídico, ocorrendo erros.

Podemo descrever que $\hat{y}$ representa o modelo predito pelo modelo e $y$ o valor real-alvo (target ou target variable).

Quando quantificamos esse erro, podemos usar ele para calcular a perda. Existem inúmeras funções de perda (loss functions), como:
 - mean-squared error (MSE)
    - Usado para dados contínuos quando a saída é um predição numérica;
    - e.g.: height, house price, temperature...
    - Aplicamos pela metade da diferença quadrática entre o valor predito e o real:
    $$
    L = \frac{1}{2}(\hat{y}-y)^2
    $$
 - Cross-entropy (logistic)
    - Usada para dados categorizáveis, quando a saída é uma probabilidade;
    - e.g.: presença de uma doença, imagem de um animal, sentimento do texto...
    - Pode ser descrita por:
    $$
    L = -(y log(\hat{y}) + (1-y)log(1-\hat{y}))
    $$

Além disso, temos a função de custo (cross function) que se relaciona com a função de perda:
$$
J = \frac{1}{n}\sum L(\hat{y}_i,y_i)
$$
É basicamente a média das perdas de muitas amostras diferentes (n).

---

### The goal of DL optimization

O objetivo principal do DL é encontrar um conjunto de pesos $\vec{w}$ que minimiza a perda $J$. Assim, usamos a loss function e, consequentemente, a cross function como critérios de otimização no treinamento. Podemos escrever essa ideia da seguinte forma:
$$
\vec{w} = \argmin(J)
$$

É interessante notar que a minimização da função de custo $L$ é possível, porém pode induzir um overfitting no modelo. Em contraponto, quando usamos a minimização da perda $J$, podemos perder parte da informação, já que estamos realizando médias. Uma forma de contornar essa situação é realizar a minimização da perda para uma quantidade $m$ de amostras menor que a quantidade total $n$ analisada.
