## The t-test

O t-test permite comparar diferentes modelos/arquiteturas a partir de uma amostra. Isso é feito pelo hipótese alternativa (Ha) e avaliamos em relação a hipótese nula (H0) considerando que ambos os modelos estão misturados e apresentam o mesmo desempenho.

O t-test é descrito por:
$$
t_k = \frac{\bar{x}-\bar{y}}{s/\sqrt{n}} = \frac{Diff.\ of\ means}{Stand.\ deviations}
$$
onde x e y representam a média, s o desvio padrão e n o número de vezes que rodamos o modelo (iterações).

Para avaliarmos esses valores t-value obtido pelo teste, usamos um distribuição esperada sob a hipótese nula (não á diferença entre as duas arquiteturas). Para isso usamos a distribuição $p-values$ sob a hipótese nula, fornecendo valores de probabilidade $P(t|H_0)$.

Dependendo do resultado do t-test e compara com o p-values e do considerado como significativo estatisticamente (0.05), podemos evidenciar se duas arquiteturas são diferentes.


```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt

# new library -> used to stats analyse
import scipy.stats as stats
```

```python
# parameters

n1 = 30 # sample in dataset 1
n2 = 40 # ...and 2
mu1 = 1 # population mean in dataset 1
mu2 = 2 # population mean in dataset 2

# generate the data
data1 = mu1 + np.random.randn(n1) # apenas criando os dados
data2 = mu2 + np.random.randn(n2)

# plot them
plt.plot(np.zeros(n1),data1,'ro',markerfacecolor='w',markersize=14)
plt.plot(np.ones(n2),data2,'bs',markerfacecolor='w',markersize=14)
plt.xlim([-1,2])
plt.xticks([0,1],labels=['Group 1','Group 2'])
plt.show()
```

```python
# t-test via stats package

# considerando as duas amostras de cada modelagem, queremos avaliar a diferença entre elas

# _ind = independent samples
t,p = stats.ttest_ind(data1,data2)
print(t)
print(p)
```

```python
# common way to show t-test results in a plot
fig = plt.figure(figsize=(10,4))
plt.rcParams.update({'font.size':12}) # change the font size

plt.plot(0+np.random.randn(n1)/15, data1, 'ro',markerfacecolor='w',markersize=14)
plt.plot(1+np.random.randn(n2)/15, data2, 'bs',markerfacecolor='w',markersize=14)
plt.xlim([-1,2])
plt.xticks([0,1],labels=['Group 1','Group 2'])

# set the title to include the t-value and p-value
plt.title(f't = {t:.2f}, p={p:.3f}')
plt.show()
```
