## Derivatives: intuition and polynomials

```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt

# sympy = simbolic math in python
import sympy as sym
import sympy.plotting.plot as symplot
```

```python
# create a symbolic variables in sympy
x = sym.symbols('x')

# create a function
fx = 2*x**2

# compute its derivative
df = sym.diff(fx,x)

# print them
print(fx)
print(df)
```

```python
# plot them
symplot(fx,(x,-4,4), title='The function')
plt.show()

symplot(df,(x,-4,4),title='Its derivative')
plt.show()
```

```python
# repeat with relu and sigmoid

# create symbolic functions
relu = sym.Max(0,x)
sigmoid = 1 / (1+sym.exp(-x))

# graph the function
p = symplot(relu,(x,-4,4),label='ReLU',show=False,line_color='blue')
p.extend( symplot(sigmoid,(x,-4,4),label='Sigmoid',show=False,line_color='red') )
p.legend = True
p.title = 'The functions'
p.show()

# graph their derivatives
p = symplot(sym.diff(relu),(x,-4,4),label='df(ReLU)',show=False,line_color='blue')
p.extend( symplot(sym.diff(sigmoid),(x,-4,4),label='df(Sigmoid)',show=False,line_color='red') )
p.legend = True
p.title = 'The functions'
p.show()
```

## Derivatives find minima

Basicamente fazemos a deriva da função f e igualamos a 0. Nisso, podemos encontrar máximos e mínimos, então verificamos o comportamento da derivada antes e depois para cada ponto cuja a derivada é 0.

Note que é possível ainda obter um ponto de derivado nula que não representa um mínimo e nem um máximo, mas sim uma transição entre eles, chamada de gradiente de desaparecimento.

## Derivatives: product and chain rules 

```python
# import libraries
import numpy as np
import sympy as sym

# make the equation look nicer
from IPython.display import display
```

```python
# create symbolic variable in sympy
x = sym.symbols('x')

# create two functions
fx = 2*x**2
gx = 4*x**3-3*x**4

# compute their individual derivatives
df = sym.diff(fx)
dg = sym.diff(gx)

# apply the product rule 'manually'
manual = df*gx + fx * dg
thewrongway = df*dg

# via sympy
viasympy = sym.diff( fx * gx)

# print everything
print('The functions: ')
display(fx)
display(gx)
print(' ')

print('Their derivatives') 
display(df)
display(dg)
print(' ')

print('Manual product rule')
display(manual)
print(' ')

print('Via sympy')
display(viasympy)
print(' ')

print('The wrong way')
display(thewrongway)
```

```python
# repeat with chain rule
fx = (x**2 + 4*x**3)**5

print('The function:')
display(fx)
print(' ')

print('Its derivative: ')
display(sym.diff(fx))
```
