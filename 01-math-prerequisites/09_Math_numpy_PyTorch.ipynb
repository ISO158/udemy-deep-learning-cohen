## Reproducible randomness via seeding

Diferentes metodologias para semear (seeding) números aleatórios. Fazer isso torna o modelo reproduzível, usando inicializações aleatórias.

```python
# import libraries
import numpy as np
import torch
```

```python
# generate a few random numbers
np.random.randn(5)
```

```python
# repeat after fixing the seed (old-but-still-widely-used method)
np.random.seed(17) # 17 é um número arbitrário, que indica o estado/posição da semeadura, nesse estado há minha semente aleatória

# a seed faz com que os valores aleatório gerados sejam os mesmos para diferentes iterações
print(np.random.randn(5))
print(np.random.randn(5))
```

## New seed mechanism in numpy (recommended)

```python
randseed1 = np.random.RandomState(17)
randseed2 = np.random.RandomState(20210530) 

# o valor do RandomState() indica um posição em que os valores random foram salvos.

print(randseed1.randn(5)) # same sequence
print(randseed2.randn(5)) # different from above. but same each time
print(randseed1.randn(5)) # same as two up
print(randseed2.randn(5)) # same as two up
print(np.random.randn(5)) # different every time
```

## Now in pytorch

```python
torch.randn(5) # tensor with standard normal variance's elements
```

```python
torch.manual_seed(17)
print(torch.randn(5))

# torch's seed doesn't spread to numpy
test = np.random.RandomState(17)
print(test.randn(5)) # fixado pelo seed

print(np.random.randn(5)) # random every time
```
