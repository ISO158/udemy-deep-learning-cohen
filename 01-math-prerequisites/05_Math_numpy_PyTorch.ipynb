## Entropy and cross-entropy

Conceito de entropia descrita por Claude Shannon. De modo geral a entropia de Shannon descreve a quantidade surpresa ou incerteza que temos sobre uma variável específica.

A entropia é máxima para probabilidade ~ 0.5 (ou seja evento imprevisíveis), a medida que a probabilidade diminui o sistema se torna mais previsível. Essa ideia é chamada de teoria da entropia da informação.

Surpresa é tratada aqui como o grau de incerteza ou o quão imprevisível é um determinado evento.

A formula da entropia é dada por:
$$
H = - \sum_{i=1}^{n} p(x_i)\ log_2(p(x_i))
$$
onde $x = data values$ e $p = probability$. A unidade utilizada é "bits" ou caso $H$ esteja descrito com log natural ($ln(p(x_i))$) a unidade é "nats".

 - Alta entropia: maior variabilidade (dispersão)
 - Menor entropia: menor variabilidade (repetição dos mesmo valores)

É interessante que apesar da entropia e a variância fornecerem informações da dispersão/variabilidade do sistema, elas são distintas. A entropia é não-linear (termo $log_2(x)$) e não faz suposições sobre a distribuição, enquanto que a variância é apropriada para distribuições aproximadamente normais.

A entropia (p) pode descreve a distribuição única de probabilidade e a 'cross-entropy' (p,q) descreve a relação entre duas distribuição de probabilidades 

$$
H = - \sum p\ log_2(p) \rightarrow H(p,q) = - \sum p\ log_2(q)
$$

## p entropy

```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# probability of an event happening
p = 0.25

# NOT the correct formula!
# essa é a probabilidade de apenas um evento ocorrer, não usamos o somatório na função H que considera qual a probabilidade do evento não ocorrer (ex: q = 1-p).
H = -(p*np.log(p))

print('Wrong entropy: ' + str(H))
```

```python
# the correct way to compute entropy
x = [0.25,0.75]

H = 0
for p in x:
    H -= p*np.log(p) # this sentence is same of H = H - p*np.log(p)

print('Correct entropy: ' + str(H))
```

```python
# also correct, written out for N = 2 events

# this system is called binary entropy - a really import lost function
H = -(p*np.log(p) + (1-p)*np.log(1-p))
print('Correct entropy',H)
# but, its resolution can be used, 'cause there's low p-values
```

## cross-entropy

```python
# note: all probs must sum to 1!
p = [1,0] # sum=1
q = [0.25,0.75] # sum=1

H = 0
for i in range(len(p)):
    H -= p[i]*np.log(q[i]) # note que p q são iterados para o mesmo i

print('Cross entropy: ', H)
```

```python
# again, we can also written for 2 elements:
H = -(p[0]*np.log(q[0]) + p[1]*np.log(q[1]))
print('Correct entropy: ',H)

# simplification, its true the p[1]*... is zero, then p[0]*np.log(q[0]) = np.log(q[0])
H = -np.log(q[0])
print('Manually simplified: ',H)
```

```python
# Let's take the same ideas using torch

import torch
import torch.nn.functional as F

# note: inputs must be Tensors
q_tensor = torch.Tensor(q)
p_tensor = torch.Tensor(p)

F.binary_cross_entropy(q_tensor,p_tensor)
#                        input , target
 # temos uma função específica no torch que trata da cross entropy. Note que precisamos usar tensores, tentar realizar a função usando listas dá erro.
```
