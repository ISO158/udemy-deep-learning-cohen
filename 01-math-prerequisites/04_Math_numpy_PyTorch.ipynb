## Logarithms

A função logarítmica é usada frequentemente em ML e otimização.

A função inversa do log(x) é a exponencial exp(x). Sendo que:
$$
log(exp(x)) = x
$$

Outro aspecto interessante é que a função log(x) é monotônica de x, isso é importante, pois minimizar x é a mesma coisa que minimizar log(x), para x>0.

O logaritmo consegue distinguir melhor valores pequenos e aproximar números espaçados.

No geral log(x) e ln(x) apresentam as mesmas propriedades, porém com slopes diferentes. Sendo que ln(x) é frequentemente usado para relacionar com $e$.

```python
# import libraries

import numpy as np
import matplotlib.pyplot as plt
```

```python
# define a set of points to evaluate
x = np.linspace(.0001,1,200)

# compute their log
logx = np.log(x)

# plot
fig = plt.figure(figsize=(10,4))

# increase font size. FYI
plt.rcParams.update({'font.size':12})

plt.plot(x,logx,'ks-',markerfacecolor='r')
plt.xlabel('x')
plt.ylabel('log(x)')
plt.show()
```

```python
# demonstration that log and exp are inverses functions

# redefine with fewer points
x = np.linspace(.0001,1,20)

# log and exp
logx = np.log(x)
expx = np.exp(x)

# the plot

plt.plot(x,x,color=[.8,.8,.8])
plt.plot(x,np.exp(logx),'o',markersize=8)
plt.plot(x,np.log(expx),'x',markersize=8)
plt.xlabel('x')
plt.ylabel('f(g(x))')
plt.legend(['unity','exp(log(x))','log(exp(x))'])
plt.show()
```
